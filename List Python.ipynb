```python
#To make a List
list1=list(("kaju","pista","usal"))
print(list1)
```

```python
#Change an item from the list.
list1[2]="bhendi chi bhaaji"
print(list1)
```

```python
#To add an item in the list.
list1.append("chapati")
print(list1)
```

```python
#To add an item at a specific point.
list1.insert(1, "kobi chi bhaaji")
print(list1)
```

```python
#To remove an item.
list1.remove("pista")
print(list1)
```

```python
#To pop an item.
list1.pop(2)
print(list1)          
```

```python
#To merge 2 lists.
list2=["falero","melody", "kaccha bite"]
list1.extend(list2)
print(list1)
```

```python
#To print a specific item from a list.
print(list1[2:5])
```

```python
#To sort items.
list3=[3,1,2]
list3.sort()
print(list3)
list3.sort(reverse=True)
print(list3)
```

```python
#CLear the list.
list3.clear()
print(list3)
```
